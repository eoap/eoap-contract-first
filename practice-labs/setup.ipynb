# Optional setup for a local notebook environment

The dev-platform workspace uses `eoepca/pde-code-server:amd64`, which already provides Bash. Workspace startup installs the course environment. Students in that workspace should go directly to [the lab index](README.md) and select **Bash**; this setup notebook is unnecessary there.

For a separate local VS Code or JupyterLab installation, open this optional notebook using an available **Python 3.12–3.14** kernel. Run the next cell to install the locked course extras and register Bash and Python kernels. It may download packages and requires Git, uv and network access. No terminal is needed.

If your instructor has provisioned the kernels, skip this notebook and select **Bash** in any module lab. A hosted Coder workspace runs commands inside its workspace pod; additional Docker, registry, Calrissian or ZOO services are used only by the marked integration sections.


```python
from pathlib import Path
import shutil
import subprocess

root = next((path for path in [Path.cwd(), *Path.cwd().parents]
             if (path / 'reference/waterbodies.cwl').is_file()), None)
if root is None:
    raise RuntimeError('Open this notebook inside the eoap-contract-first checkout')
uv = shutil.which('uv')
if uv is None:
    raise RuntimeError('The course workspace must provide uv; ask the instructor to provision it')
subprocess.run([uv, 'sync', '--locked', '--extra', 'tooling', '--extra', 'dev', '--extra', 'notebook'], cwd=root, check=True)
subprocess.run([uv, 'run', '--no-sync', 'bash', 'scripts/install_notebook_kernels.sh', '--user'], cwd=root, check=True)
print('Setup complete. Reopen a module lab, refresh the kernel picker and select Bash.')
```
